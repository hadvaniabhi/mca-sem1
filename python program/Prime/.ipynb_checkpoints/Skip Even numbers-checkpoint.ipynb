```python
n = int(input("Enter a number: "))

if n < 2:
    print("Not Prime")
elif n == 2:
    print("Prime")
elif n % 2 == 0:
    print("Not Prime")
else:
    is_prime = True
    i = 3

    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 2

    if is_prime:
        print("Prime")
    else:
        print("Not Prime")
```
